```python
from langchain_core.tools import tool
```

kono tool banaite hoile
- @tool diye tool decorator use korte hoy 
- funtion banate hoy jekhane make sure korte hoy argument types
- """ """ bhetore work (dock string) define korte hoy

```python
@tool 
def add_numbers(a: int, b: int):
    """
        it will return the sum of two numbers

        Args:
            a : Number one
            b : Number two
    
    """

    #args decide kore kono function kon kon argument receive hobe

    return a+b

@tool 
def mul_numbers(a: int, b: int):
    """
        it will return the product of two numbers

        Args:
            a : Number one
            b : Number two
    
    """

    #args decide kore kono function kon kon argument receive hobe

    return a+b
```

```python
add_numbers.invoke({"a": 12, "b": 8}) #jehetu duita arguments pass korbo tai dictionary use kora hoise
```

```python
from langchain_groq import ChatGroq
from langchain.agents import create_agent #create agent LLM k agentic work er shathe connect kore
```

```python
llm = ChatGroq(model="openai/gpt-oss-120b")
```

an agent must contain two things
- model jeta use korbo
- multiple tools
- system prompt o thakbe

```python
agent = create_agent(
    model = llm,
    tools = [add_numbers, mul_numbers],
    system_prompt = "You are a math teacher, and you always use tools for calculation"
)
```

```python
res = agent.invoke({"messages":[{"role": "user", "content": "What is 8+3*10?"}]})
print(res["messages"][-1].content)
```
