### Problem: ekhane duita string and ekta int input neya uchit but keu jodi int e string dey tao she nibe. which should not be right. 

```python
class User:
    name = "" 
    email = ""
    age = 0
```

```python
user = User()
user.name = "Rafin"
user.email = "email@gmail.com"
user.age = "My age is: 23"

print(user)
```

### Solution: Pydantic 

```python
from pydantic import BaseModel, Field
from typing import List, Annotated
```

```python
class User(BaseModel):
    name : str
    age : int
```

##### ekhane jodi age e jodi string dei error khabe

```python
user = User(name="Rafin", age=23)
user
```

##### Field Validation

```python
class Product(BaseModel):
    title : str = Field(max_length = 10) # size define er jonno use hoy
    des : str = Field(max_length = 50)
    price : int = Field(gt = 10) # gt means greater than 10 
```

```python
product = Product(title="Nothing", des = "more unique than others", price = 35000)
product2 = Product(title="Samsung", des = "more heavy than others", price = 150000)
product3 = Product(title="iPhone", des = "more luxurious than others", price = 35000)
product
```

```python
class Order(BaseModel):
    id : str = Field(min_length=5, max_length = 10) # size define er jonno use hoy
    product : Product # Product type er object like str and int
    
```

```python
order = Order(id="55xAE3", product=product)
order
```

```python
order.product.title
```

```python
class Carts(BaseModel):
    id : str = Field(min_length=5, max_length = 10) # size define er jonno use hoy
    items : List[Product] # Product object er list jeno product1, product2 nite pari
    comment : str  = Field(default="No Comment") # optional jeta thakteo pare nao thakte pare 
```

```python
cart  = Carts(id = "fxxyxxxl", items=[product, product2, product3], comment="This is my cart")
cart
```

```python
for product in cart.items:
    print(product.title)
```

##### Annotated use kora hoy to use  a same thing many times

```python
commonValidator = Annotated[List[Product], Field(default=[])]
```

```python
class Carts(BaseModel):
    id : str = Field(min_length=5, max_length = 10) # size define er jonno use hoy
    items : commonValidator
    comment : str  = Field(default="No Comment") # optional jeta thakteo pare nao thakte pare 
```

```python
class Order(BaseModel):
    
    product : commonValidator 
```

```python
ord = Order()
ord
```
