#### Structured Output

```python
from pydantic import BaseModel, Field
#BaseModel help kore datatype define korte
#Field help kore datatype er description bujhate

from typing import List
#Object variable type banaite help kore

class ResponseStructure(BaseModel):
    name:str = Field(description="Name of the user")
    email:str = Field(description="Email Address of the user")
    age:int = Field(description="Age of the user")




structured_llm = llm.with_structured_output(ResponseStructure)
#LLM k structured output provide kora jeta holo "ResponseStructure"
```

```python
res = structured_llm.invoke(f"Please give me only name age and email for the text: {text}")
```

```python
res.age
```

```python
res.email
```

```python
class Movies(BaseModel):
    title:str = Field(description="Movie title")
    year:int = Field(description="Movie released Year")


class AllMovies(BaseModel):
    movies:List[Movies]
```

```python
structured_llm_movie = llm.with_structured_output(AllMovies)
```

```python
res = structured_llm_movie.invoke("Give me Five recent Sci-fi movies")
res
```
