```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_groq import ChatGroq
from langchain.agents import create_agent
from langchain_community.utilities import GoogleSerperAPIWrapper
from langgraph.checkpoint.memory import InMemorySaver 
```

```python
llm = ChatGroq(model="openai/gpt-oss-20b")
search = GoogleSerperAPIWrapper()
```

```python
agent = create_agent(
    model=llm, 
    tools=[search.run],
    checkpointer=InMemorySaver(),
    system_prompt="You are an agent and You can search for any questions on Google"
)
```

```python
question = "What won the women's world cup 2025"
```

Agent er response nite hole list er moddhe key value "message" object niye dite hobe. template: agent.invoke({"messages":[{"role": "user", "content": question}]})

```python
res = agent.invoke(
    {"messages":[{"role": "user", "content": question}]},
    {"configurable": {"thread_id": "1"}}, # thread is changable 
    )
```
