```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_groq import ChatGroq
```

```python
llm = ChatGroq(
    model="qwen/qwen3-32b"
)
```

```python
prompts = [
    {"role": "system", "content": "You are a Python expert."},
    {"role": "user", "content": "What is the difference between a list and a tuple in Python?"}
]

res =  llm.invoke(prompts)
print(res)
```

### Prompt Template

```python
from langchain_core.prompts import ChatPromptTemplate
```

```python
llm = ChatGroq( model="meta-llama/llama-4-scout-17b-16e-instruct")
```

```python
prompts = ChatPromptTemplate.from_messages(
    [
        {"role": "system", "content": "You are a Python expert."},
        {"role": "user", "content": "{question}"}
    ]
)
```

```python
final_prompt = prompts.format_prompt(
    question="Why python is time expensive?"
    )
final_prompt
```

```python
res  = llm.invoke(final_prompt)
print(res.content)
```

```python
prompts.invoke({"question": "What is python?"})
```

### Chaining

```python
res =  llm.invoke(prompts.invoke({"question": "What is python?"}))
print(res.content)
```

```python
chains = prompts | llm
chains
```

```python
res = chains.invoke({"question": "Who invent3ed python?"})
print(res.content)  
```

```python
from langchain_core.output_parsers import StrOutputParser

out = StrOutputParser()
```

```python
chains = prompts | llm | out
res = chains.invoke({"question": "Who invent3ed python?"})
print(res) 
```
