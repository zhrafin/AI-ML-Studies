```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
text = "Today, I want to speak from the heart about three pillars " \
"that define the future of our nation: our youth, our education system, and our farmers."
```

chunk_size = 50: 50 size er chunk hobe, chunk_overlap = 20: ager chunk er 20 next chunk e thakbe to make a relation

```python
splitter = RecursiveCharacterTextSplitter(chunk_size = 500, chunk_overlap = 200)
splitted_texts = splitter.split_text(text)
splitted_texts
```

```python
from langchain_community.document_loaders import PyPDFLoader, TextLoader
file_path = "../data/speech.txt"
loader = TextLoader(file_path)
texts = loader.load()
```
