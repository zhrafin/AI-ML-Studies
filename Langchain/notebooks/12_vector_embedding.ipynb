```python
from dotenv import load_dotenv
load_dotenv()
```

```python
doccuments = ["Upgrade government schools with smart classrooms",
              "Improve teacher training",
              "Expand scholarships for higher education",
              "Promote technical and vocational courses",
              "Ensure every child gets a fair opportunity to learn and grow"]
```

normal query er jonno

```python
from langchain_ollama import OllamaEmbeddings

embeddings = OllamaEmbeddings(model="llama3")
resultQ = embeddings.embed_query("hello world")
print(len(resultQ))
```

document text er jonno

```python
from langchain_ollama import OllamaEmbeddings

embeddings = OllamaEmbeddings(model="llama3")
resultD = embeddings.embed_documents(doccuments)
print(len(resultD))
print(resultD)
```

#### Store vector embedding using Chroma

```python
from langchain_chroma import Chroma
```

```python
vector_store = Chroma.from_texts(
    texts=doccuments,
    embedding=embeddings,
    persist_directory="./vector_db"
)
```

```python
query = "Machine Learning"
result = vector_store.similarity_search(query=query, k=2)
```
