```python
from dotenv import load_dotenv
load_dotenv()
```

```python
doccuments = ["Upgrade government schools with smart classrooms",
              "Improve teacher training",
              "Expand scholarships for higher education",
              "Promote technical and vocational courses",
              "Ensure every child gets a fair opportunity to learn and grow"]
```

```python
from langchain_ollama import OllamaEmbeddings

embeddings = OllamaEmbeddings(model="llama3")
resultQ = embeddings.embed_query("hello world")
print(len(resultQ))
```

```python
from langchain_ollama import OllamaEmbeddings

embeddings = OllamaEmbeddings(model="llama3")
resultD = embeddings.embed_documents(doccuments)
print(len(resultD))
print(resultD)
```
